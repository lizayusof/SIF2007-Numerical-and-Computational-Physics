## Importing the plotting library

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
Path("figures").mkdir(exist_ok=True)
```

## A first plot: sine

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)
y = np.sin(x)
plt.figure(figsize=(7, 4))
plt.plot(x, y)
plt.xlabel("x (rad)")
plt.ylabel("sin(x)")
plt.title("Sine function")
plt.grid(alpha=0.3)
plt.show()
plt.close()
```

## Plotting sine and cosine together

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)
plt.figure(figsize=(7, 4))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), linestyle="--", label="cos(x)")
plt.xlabel("x (rad)")
plt.ylabel("Function value")
plt.title("Sine and cosine")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("figures/sine_cosine.png", dpi=300,
            bbox_inches="tight")
plt.savefig("figures/sine_cosine.pdf", bbox_inches="tight")
plt.show()
plt.close()
```

## Changing the appearance

```python
x_points = np.linspace(0, 2 * np.pi, 12)
plt.figure(figsize=(7, 4))
plt.plot(x_points, np.sin(x_points), color="purple",
         marker="o", linestyle="--", linewidth=1.5,
         label="Sampled sine values")
plt.xlabel("x (rad)")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
plt.close()
```

## Optional: a simple animation

```python
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML, display

x_anim = np.linspace(0, 2 * np.pi, 60)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x_anim, np.sin(x_anim), color="0.7")
point, = ax.plot([], [], "ro")
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.2, 1.2)
ax.set_xlabel("x (rad)")
ax.set_ylabel("sin(x)")

def update(frame):
    point.set_data([x_anim[frame]], [np.sin(x_anim[frame])])
    return (point,)

animation = FuncAnimation(fig, update, frames=len(x_anim),
                          interval=80, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
# Optional GIF file; requires Pillow:
# animation.save("figures/sine_animation.gif",
#                writer=PillowWriter(fps=12))
```
