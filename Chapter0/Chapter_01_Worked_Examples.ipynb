## Lists and simple output formatting

```python
values = [10, 20, 30]
print(values[0])      # 10
print(values[-1])     # 30
print(len(values))    # 3
values.append(40)
print(values)        # [10, 20, 30, 40]

mass_kg = 0.5
speed_m_s = 4.0
energy_J = 0.5 * mass_kg * speed_m_s**2
print(f"Kinetic energy = {energy_J:.2f} J")
```
