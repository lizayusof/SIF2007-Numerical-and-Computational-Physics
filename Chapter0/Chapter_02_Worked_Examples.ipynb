## Selection: if, else and elif

## The if statement

```python
a = 5
b = 3
if a > b:
    print("a is greater than b")
```

## The if--else statement

```python
a = 2
b = 3
if a > b:
    print("a is greater than b")
else:
    print("a is not greater than b")
```

## A nested if statement

```python
x = 5
if x > 0:
    if x < 10:
        print("x is positive and less than 10")
    else:
        print("x is at least 10")
else:
    print("x is zero or negative")
```

## The if--elif--else statement

```python
x = -2
if x > 0:
    print("x is positive")
elif x == 0:
    print("x is zero")
else:
    print("x is negative")
```

## Comparison and logical operators

```python
x = 5
print(x >= 2)            # True
print(x <= 4)            # False
print(x > 0 and x < 10)  # True
print(x < 0 or x > 10)   # False
print(not (x == 5))      # False
```

## Repetition: for loops

```python
for i in range(5):
    print(i)

for value in [1, 2, 3]:
    print(value**2)
```

## Repetition: while loops

```python
i = 0
while i < 5:
    print(i)
    i = i + 1
```

## Functions

```python
def greeting():
    print("Creating our first function")

greeting()
```

```python
def polynomial(x):
    return x**3 + 2 * x**2 + 5

print(polynomial(0))     # 5
print(polynomial(2))     # 21
result = polynomial(3)
print(result)            # 50
```
